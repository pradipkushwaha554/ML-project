```python
import ast

def convert_to_list(x):
    list=[]
    for i in ast.literal_eval(x):
        list.append(i['name'])
    return list
```

```python
convert_to_list('[{"id": 28, "name": "Action"}, {"id": 12, "name": "Adventure"}, {"id": 14, "name": "Fantasy"}, {"id": 878, "name": "Science Fiction"}]')
```

```python
movies['genres']=movies['genres'].apply(convert_to_list)
movies.head(5)
```

```python
movies.iloc[0].keywords
```

```python
def convert_to_keywords(x):
    list = []
    for i in ast.literal_eval(x):
        list.append(i['name'])
    return list
```

```python
convert_to_keywords(movies.iloc[0].keywords)
```

```python
movies['keywords']=movies['keywords'].apply(convert_to_keywords)
```

```python
movies.head(5)
```

```python
def convert_to_cast(x):
    list = []
    counter=0
    for i in ast.literal_eval(x):
        if counter!=3:
            list.append(i['name'])
            counter+=1
        else:
            break
    return list
```

```python
convert_to_cast(movies.iloc[0].cast)
```

```python
movies['cast']=movies['cast'].apply(convert_to_cast)
```

```python
movies.head(5)
```

```python
def fetch_to_director(x):
    list=[]
    for i in ast.literal_eval(x):
        if i['job'] == 'Director':
            list.append(i['name'])
    return list
```

```python
fetch_to_director(movies['crew'].iloc[0])
```

```python
movies['crew']=movies['crew'].apply(fetch_to_director)
```

```python
movies.head(5)
```

```python
movies['overview']=movies['overview'].apply(lambda x: x.split())
```

```python
movies['overview'].head(5)
```

```python
movies['genres']=movies['genres'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['keywords']=movies['keywords'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['cast']=movies['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
movies['crew']=movies['crew'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
movies.head(5)
```

```python
movies['tag']=movies['overview']+movies['genres']+movies['keywords']+movies['cast']+movies['crew']
```

```python
new_df=movies.drop(['overview', 'genres', 'keywords', 'cast', 'crew'], axis=1)
```

```python
new_df.head()
```

```python
new_df['tag']=new_df['tag'].apply(lambda x: " ".join(x))
```

```python
new_df.head(5)
```

```python
new_df['tag'][0]
```

```python
new_df['tag']=new_df['tag'].apply(lambda x: x.lower())
```

```python
new_df.head(5)
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    y=[]
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
```

```python
new_df['tag']=new_df['tag'].apply(stem)
```

```python
# vectorization 
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000, stop_words='english')
vectors=cv.fit_transform(new_df['tag']).toarray()
```

```python
vectors[0]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity=cosine_similarity(vectors)
```

```python
list(enumerate(similarity[0]))
```

```python
def recommend(movie_name):
    idx = new_df[new_df['title'] == movie_name].index[0]
    distances = sorted(enumerate(similarity[idx]), reverse=True, key=lambda x: x[1])
    recommended_movies = []
    for i in distances[1:6]:
        recommended_movies.append(new_df.iloc[i[0]]['title'])
    return recommended_movies 
```

```python
recommend('Avatar')
```
